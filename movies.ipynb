```python
## Clean the Data
```

```python
cols =['id', 'title', 'release_date', 'runtime', 'budget', 'revenue', 'genres', 'vote_average', 'vote_count', 'original_language']
movies = movies[cols].copy()
```

```python
# Fix impossible zeros
movies[['budget', 'revenue']] = movies[['budget', 'revenue']].replace(0, np.nan)
```

```python
# handle missing runtimes
movies = movies.dropna(subset=['runtime'])
```

```python
# convert dates
movies['release_date'] = pd.to_datetime(movies['release_date'])
movies['year'] = movies['release_date'].dt.year
movies['decade'] = (movies['year'] // 10) * 10
```

```python
import ast

# Parse the genres column
movies['genres'] = movies['genres'].apply(ast.literal_eval)
movies['genre_names'] = movies['genres'].apply(lambda lst: [g['name'] for g
in lst])
```

```python
# Remove duplicates
movies = movies.drop_duplicates(subset='id')
```

```python
# The reliable-rating subset is created after runtime categories are added.
```

```python
movies.info()
```

```python
# Feature Engineering

# Profit and return on investment
movies['profit'] = movies['revenue'] - movies['budget']
movies['roi'] = movies['profit'] / movies['budget']
```

```python
# Runtime categories
movies['length'] = pd.cut(movies['runtime'], bins=[0, 90, 120, 500], labels=['short', 'medium', 'long'])

# Keep this subset after feature engineering so it includes the `length` column.
rated = movies[movies['vote_count'] >= 100].copy()
```

```python
# Weighted rating
C = movies['vote_average'].mean()
m = movies['vote_count'].quantile(0.80)
v = movies['vote_count']
R = movies['vote_average']
movies['weighted_rating'] = (v / (v + m)) * R + (m / (v + m)) * C
```

```python
# Genres 
# Explode the genre list
genre_df = movies.explode("genre_names")
genre_df.rename(columns={'genre_names': 'genre'}, inplace=True)
genre_df[['title', 'genre']].head(10)
```

```python
# Count movies per genre
movies_per_genre = genre_df['genre'].value_counts()
movies_per_genre
```

```python
# Average rating per genre with groupby
genre_stats = (genre_df[genre_df['vote_count'] >= 100]
        .groupby('genre')
        .agg(movie_count=('title', 'count'),
        avg_rating=('vote_average', 'mean'),
        avg_runtime=('runtime', 'mean'),
        median_revenue=('revenue', 'median'))
        .sort_values('avg_rating', ascending=False))
genre_stats
```

## Perfom Analisis 




```python
# Check the linear relationship between runtime and rating for movies with enough votes.
rated[["runtime", "vote_average"]].corr()

# Compare the rating distribution across the short, medium, and long categories.
rated.groupby("length", observed=True)["vote_average"].agg(["count", "mean", "median"])
```

```python
# Keep only movies with known budget and revenue so profit comparisons are meaningful.
budget_data = movies.dropna(subset=["budget", "revenue"]).copy()
budget_data[["budget", "revenue", "vote_average", "profit"]].corr()

# Split budgets into four similarly sized groups and compare their outcomes.
budget_data["budget_quartile"] = pd.qcut(
    budget_data["budget"],
    q=4,
    labels=["low", "mid-low", "mid-high", "high"],
)
budget_data.groupby("budget_quartile", observed=True).agg(
    movie_count=("title", "count"),
    median_budget=("budget", "median"),
    mean_rating=("vote_average", "mean"),
    median_profit=("profit", "median"),
)
```

```python
# Use the reliable-rating subset and keep counts beside averages to show sample size.
rated.groupby("decade")["vote_average"].agg(["count", "mean"])
```

```python
# Group by calendar month; median revenue is less affected by blockbuster outliers.
movies.groupby(movies["release_date"].dt.month)["revenue"].median()
```

```python
# Add decade to the exploded table so each movie-genre row can be grouped by decade.
genre_df["decade"] = (genre_df["year"] // 10) * 10

# Rows are genres and columns are decades; cells contain average ratings.
genre_decade_ratings = pd.pivot_table(
    genre_df[genre_df["vote_count"] >= 100],
    values="vote_average",
    index="genre",
    columns="decade",
    aggfunc="mean",
)
genre_decade_ratings
```

```python
# The 1.5 x IQR rule flags values far outside the middle half of runtimes.
q1, q3 = rated["runtime"].quantile([0.25, 0.75])
iqr = q3 - q1

# Parentheses are needed around each comparison when combining masks with |.
outliers = rated[
    (rated["runtime"] < q1 - 1.5 * iqr)
    | (rated["runtime"] > q3 + 1.5 * iqr)
]
outliers[["title", "runtime"]].sort_values("runtime")
```

## Merge the Credit Data

```python
# Inspect the credits
credits.head()
```

```python
credits.columns
```

```python
# Extract the director
credits['crew'] = credits['crew'].apply(ast.literal_eval)

def get_director(crew_list):
    for person in crew_list:
        if person['job'] == 'Director':
            return person['name']
    return np.nan

credits['director'] = credits['crew'].apply(get_director)
```

```python
# merge
movies = movies.merge(credits[['movie_id', 'director']],
                      left_on='id', right_on='movie_id', how='left')
```

```python
# Director analysis
director_stats = (movies[movies['vote_count'] >= 100]
                                .groupby('director')
                                .agg(films=('title', 'count'),
                                avg_rating=('vote_average', 'mean'),
                                rating_std=('vote_average', 'std'))
                                .query('films >= 4')
                                .sort_values('avg_rating', ascending=False))
director_stats.head(15)
```

```python
# Explore lead actors
# Parse the cast text into Python lists of dictionaries.
credits["cast"] = credits["cast"].apply(ast.literal_eval)

# Keep the first three billed performers for each movie.
credits["lead_actors"] = credits["cast"].apply(
    lambda cast_list: [person["name"] for person in cast_list[:3]]
)

# Explode the actor list so there is one row per movie-actor combination.
actor_movies = credits[["movie_id", "lead_actors"]].explode("lead_actors")
actor_movies = actor_movies.rename(columns={"lead_actors": "actor"})

# Add movie ratings and vote counts to each actor row.
actor_movies = actor_movies.merge(
    movies[["id", "title", "vote_average", "vote_count"]],
    left_on="movie_id",
    right_on="id",
    how="left",
)

# Use only movies with at least 100 votes, then require four films per actor
# so the ranking is less likely to be driven by a single appearance.
actor_stats = (
    actor_movies[actor_movies["vote_count"] >= 100]
    .groupby("actor")
    .agg(
        films=("title", "nunique"),
        avg_rating=("vote_average", "mean"),
        median_rating=("vote_average", "median"),
    )
    .query("films >= 4")
    .sort_values("avg_rating", ascending=False)
)

actor_stats.head(15)
```
